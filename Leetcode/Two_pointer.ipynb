```python
nums = [2, 7, 11, 15]
target = 9


def two_sum(nums, target):
    left = 0
    right = len(nums)-1
    while left<right:
        current = nums[left]+nums[right]
        if current==target:
            return (left+1, right+1)
        elif current >target:
            right-=1
        else:
            left+=1
print(two_sum(nums, target))
```

```python
st = "madam"


def check_palin(st):
    left = 0
    right = len(st)-1
    while left<right:
        if st[left]!=st[right]:
            return "Not a Palindrome"
        left+=1
        right-=1

    return "Palindrome"

print(check_palin(st))
    
```

```python
l1 = ["h","e","l","l","o"]

def reverse_list(l1):
    left = 0
    right = len(l1)-1
    while left<right:
        l1[left], l1[right] = l1[right], l1[left]
        left+=1
        right-=1
reverse_list(l1)
print(l1)
```

```python
nums = [1,1,2,2,3]

def remove_duplicates(nums):
    p1 = 0
    p2 = 1

    while p2<len(nums):
        if nums[p1]!=nums[p2]:
            p1+=1
            
            nums[p1]=nums[p2]
        p2+=1
    return p1+1

print(remove_duplicates(nums))

print(nums)

```

```python
# Container With Most Water

height = [1,8,6,2,5,4,8,3,7]

def max_area(height):
    left = 0
    right = len(height)-1
    max_area = 0
    while left<right:

        width = right-left

        current_area = width * min(height[left], height[right])

        max_area = max(current_area, max_area)

        if height[left]<height[right]:
            left+=1
        else:
            right-=1

    return max_area

print(max_area(height))
```
