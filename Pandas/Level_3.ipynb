Level 3 - Slicing (.loc/.iloc)

```python
# create a dataframe of below data

# Name      Dept      Salary    Age
# John      IT        70000      25
# Sam       HR        50000      31
# Lisa      IT        90000      29
# David     Sales     60000      27
# Sara      HR        55000      35

import pandas as pd


data = {
        'Name': ['John', 'Sam', 'Lisa', 'David', 'Sara'],
        'Dept': ['IT', 'HR', 'IT', 'Sales', 'HR'],
        'Salary': [70000, 50000, 90000, 60000, 55000],
        'Age': [25, 31, 29, 27, 35]
    }
df = pd.DataFrame(data)


```

```python
df
```

```python
# 31

# First three rows.

df[:3]
```

```python
# 32

# Last two rows.

df[-2:]
```

```python
# 33

# Only Name and Salary.

df.iloc[::, ::2]

df[['Name', 'Salary']]
```

```python
# 34

# Rows 3-6.

df[3:6]
```

```python
# 35

# Rows where Age >30 and select Salary only.

df[df['Age']>30].loc[::, 'Salary']
df.loc[df['Age'] > 30, 'Salary']
```

```python
# 36

# Every alternate row.

df.iloc[::2]
```

```python
# 37

# Reverse dataframe.

df[::-1]
```

```python
# 38

# Last column only.

df.iloc[::, -1]
```

```python
# 39

# First row all columns except last.

df.iloc[0:1, :-1]
```

```python
# 40

# Select

# rows 2 to 8

# columns 1 to 3


df.iloc[2:9,1:4]
```
