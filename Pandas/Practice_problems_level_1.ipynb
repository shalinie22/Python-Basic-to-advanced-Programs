```python
import pandas as pd
```

```python
# Problem 1

# Create the following DataFrame.

S = {"Name":['John','Sam','Lisa'],
     "Age": [23,25,21],
     "Salary":[45000, 52000, 48000]}

df = pd.DataFrame(S)

df
```

```python
# Problem 2

# Print

# first 2 rows
# last row
# column names
# datatype
# shape

print(df.head(2))
print(df.tail(1))
print(df.columns)
print(df.dtypes)
print(df.shape)
```

```python
# Problem 3

# Change

# Salary

# to

# MonthlySalary

df.rename({'Salary':'MonthlySalary'}, axis=1, inplace=True)

df
```

```python
# Problem 4

# Add new column

# Experience

df['Experience'] = [3,5,4]

df
```

```python
# Problem 5

# Delete

# Experience


df.drop(columns=['Experience'], axis =1, inplace=True)

df
```

```python
# Problem 6

# Set

# Name

# as index.


df.set_index('Name', inplace=True)
df
```

```python
# Problem 7

# Reset the index.

df.reset_index(inplace=True)
df
```
