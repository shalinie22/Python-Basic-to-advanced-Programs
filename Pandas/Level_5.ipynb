Level 5 - Apply()

```python
import pandas as pd
```

```python
# create a dataframe of below data

# Name      Dept      Salary    Age
# John      IT        70000      25
# Sam       HR        50000      31
# Lisa      IT        90000      29
# David     Sales     60000      27
# Sara      HR        55000      35
# 


data = {
        'Name': ['John', 'Sam', 'Lisa', 'David', 'Sara'],
        'Dept': ['IT', 'HR', 'IT', 'Sales', 'HR'],
        'Salary': [70000, 50000, 90000, 60000, 55000],
        'Age': [25, 31, 29, 27, 35]
    }
df = pd.DataFrame(data)
```

```python
# 46

# Increase salary by 10%.
df['Salary'].apply(lambda x: x+(x*0.1))


```

```python
df
```

```python
# 47

# Convert to monthly salary.

df['Monthly_Salary']=df['Salary'].apply(lambda x: x//12)

```

```python
df
```

```python
# 48

# Return

# High

# Medium

# Low

# based on salary.

df['Category'] = df['Salary'].apply(lambda x: "High" if x>=80000 else ("Low" if x<=50000 else "Medium") )
 
```

```python
df
```

```python
# Problem 49 - Extract First Letter

df['Name'].apply(lambda x: x[0])
```

```python
# Problem 50 - Convert Names to Uppercase

df['Name']=df['Name'].apply(lambda x:x.upper())
```

```python
df
```

```python
# Problem 53 - Convert Gender

df['Gender'] = ["Male", "Male", "Female", "Male", "Female"]

df
```

```python
df['Gender'] = df['Gender'].apply(lambda x:"M" if x=="Male" else "F")
```

```python
df
```

```python
# Problem 51 - Calculate Tax

df['Tax'] = df['Salary'].apply(lambda x: x*0.1 if x<70000 else x*0.2)

df
```

```python
# Problem 52 - Age Category

df['Age_Category'] = df["Age"].apply(lambda x: "Young" if x<30 else ("Senior" if x>50 else "Adult"))

df
```

```python
# Problem 54 - Calculate Bonus

df['Bonus'] = df['Salary'].apply(lambda x: x*0.2 if x>80000 else (x*0.10 if x<60000 else x*0.15))

df
```

```python
# Problem 55 - Multiple Column Calculation

df['Total_income'] = df.apply(lambda x: x['Salary'] + x['Bonus'], axis=1)

df
```
