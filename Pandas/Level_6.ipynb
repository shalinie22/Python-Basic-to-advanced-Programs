Level 7 - Replace()

```python
import pandas as pd
```

```python
# create a dataframe of below data

# Name      Dept      Salary    Age
# John      IT        70000      25
# Sam       HR        50000      31
# Lisa      IT        90000      29
# David     Sales     60000      27
# Sara      HR        55000      35



data = {
        'Name': ['John', 'Sam', 'Lisa', 'David', 'Sara'],
        'Dept': ['IT', 'HR', 'IT', 'Sales', 'HR'],
        'Salary': [70000, 50000, 90000, 60000, 55000],
        'Age': [25, 31, 29, 27, 35]
    }
df = pd.DataFrame(data)
```

```python
df
```

```python
# 61

# Replace

# Male

# Female

df['Gender'] = ["Male", "Male", "Female", "Male", "Female"]

df['Gender'] = df['Gender'].replace({"Male":"M","Female":"F"})
```

```python
df
```

```python
# 62

# Replace

# 0

# NaN

df['Dept'] = df['Dept'].map(lambda x: 0 if x=="IT" else x)

df
```

```python
import numpy as np
```

```python
df = df.replace(0,np.nan) 
df
```

```python
# 63

# Replace multiple values.

df = df.replace(["John",90000],["JOHN",100000])

df
```

```python
# 64

# Replace entire department.

df['Dept'].replace({
    np.nan:"Information Technology",
    "HR":"Human Resource",
    "Sales" : "Sales Management"
}, inplace=True)

df
```

```python
# 65

# Regex replace.

df['Name'] = df['Name'].replace("$","1", regex=True)

df
```
