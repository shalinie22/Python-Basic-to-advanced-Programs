```python
import pandas as pd
import numpy as np



```

```python
# create a dataframe of below data

# Name      Dept      Salary    Age
# John      IT        70000      25
# Sam       HR        50000      31
# Lisa      IT        90000      29
# David     Sales     60000      27
# Sara      HR        55000      35



data = {
        'Name': ['John', 'Sam', 'Lisa', 'David', 'Sara'],
        'Dept': ['IT', 'HR', 'IT', 'Sales', 'HR'],
        'Salary': [70000, 50000, 90000, 60000, 55000],
        'Age': [25, 31, 29, 27, 35]
    }
df = pd.DataFrame(data)


```

```python
df
```

```python
# Beginner
# 11

# Find employees whose salary > 60000.


df[df['Salary']>60000]
```

```python
# 12

# Age < 30.

df[df['Age']<30]
```

```python
# 13

# Department == IT

df[df['Dept']=="IT"]
```

```python
# 14

# Department != HR

df[df['Dept']!="HR"]
```

```python
# 15

# Salary between 50000 and 80000


df[(df['Salary']>50000) & (df['Salary']<80000)]
```

```python
# Intermediate
# 16

# IT employees earning >70000


df[(df['Dept']=="IT") & (df['Salary']>70000)]
```

```python
# 17

# HR employees older than 30


df[(df['Dept']=="HR") & (df['Age']>30)]
```

```python
# 18

# Salary >60000 OR age <30

df[(df['Salary']>60000) | (df['Age']<30)]
```

```python
# 19

# Salary >60000 AND age<30

df[(df['Salary']>60000) & (df['Age']<30)]
```

```python
# 20

# Employees NOT in Sales.

df[df['Dept']!="Sales"]
```

```python
# 21

# Age is one of

# 25
# 29
# 35


df[df['Age'].isin([25,29,35])]
```

```python
# 22

# Department starts with "S"

df[df['Dept'].str.startswith("S")]
```

```python
# 23

# Name ends with "a"


df[df['Name'].str.endswith("a")]
```

```python
# 24

# Names containing "a"

df[df['Name'].str.contains('a')]
```

```python
# 25

# Filter multiple departments

# IT
# HR

df[(df['Dept']=="IT")|(df['Dept']=="HR")]
```

```python
df
```

```python
# Advanced
# 26

# Top 3 highest salary employees.

df.sort_values(by='Salary', ascending=False).head(3)
```

```python
# 27

# Second highest salary.

second = df['Salary'].drop_duplicates().sort_values(ascending=False).iloc[1]

df[df['Salary']==second]
```

```python
# 28

# Youngest employee.

df[df['Age'] == df['Age'].sort_values().iloc[0]]
```

```python
# 29

# Oldest employee.

df[df['Age'] == df['Age'].sort_values().iloc[-1]]
```

```python
# 30

# Employees having duplicate salary.

df[df['Salary'].duplicated(keep=False)]
```
