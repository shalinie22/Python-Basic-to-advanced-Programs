Matplotlib VS Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

df = pd.DataFrame({
    "Age":[20,22,25,30,35],
    "Salary":[30000,35000,45000,55000,70000]
})
```

```python
df
```

Line Plot

```python
plt.plot(df['Age'], df['Salary'])
```

```python
sns.lineplot(data=df, x='Age', y='Salary')
```

```python
# Bar Plot

plt.bar(df['Age'], df['Salary'])
```

```python
sns.barplot(data=df, x='Age', y='Salary')
```

```python
# Scatter Plot

plt.scatter(df['Age'], df['Salary'])
```

```python
sns.scatterplot(data=df, x='Age', y='Salary')
```

```python
plt.hist(df['Salary'])
```

```python
sns.histplot(df['Salary'])
```
